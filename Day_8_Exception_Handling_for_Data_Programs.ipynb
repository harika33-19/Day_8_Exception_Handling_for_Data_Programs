```python
def safe_convert(value):
    try:
        number = float(value)
        return number

    except ValueError:
        return None


values = ["25", "50.5", "abc", "100"]

for value in values:
    result = safe_convert(value)

    if result is not None:
        print(value, "->", result)
    else:
        print(value, "-> Invalid value")
```

```python
marks = ["85", "90", "abc", "76", "invalid"]

valid_marks = []

for mark in marks:
    try:
        mark = float(mark)
        valid_marks.append(mark)

    except ValueError:
        print("Invalid mark:", mark)

print("Valid marks:", valid_marks)
```

```python
student_data = ["85", "90", "75", "abc", "88"]

total = 0
count = 0

for value in student_data:
    try:
        mark = float(value)
        total += mark
        count += 1

    except ValueError:
        print("Skipping invalid value:", value)

if count > 0:
    average = total / count
    print("Average marks:", average)
else:
    print("No valid marks found.")
```
